```
import torch
import matplotlib.pyplot as plt
```

```
bw_img = torch.randint(2,(50,50))
print(bw_img.shape)
plt.figure(figsize=(3, 3))
plt.imshow(bw_img,cmap = 'gray')
```

```
gray_img = torch.randint(256,(50,50))
print(gray_img.shape)
plt.figure(figsize=(3, 3))
plt.imshow(gray_img,cmap = 'gray')
```

```
img = torch.randint(255,(3,256,256))
print(img.shape)
img = img.permute(1,2,0)
print(img.shape)
plt.figure(figsize=(3, 3))
plt.imshow(img)
```

```
rimg = torch.zeros_like(img,dtype = img.dtype)
gimg = torch.zeros_like(img,dtype = img.dtype)
bimg = torch.zeros_like(img,dtype = img.dtype)

rimg[:,:, 0 ] = img[:,:,0]
gimg[:,:, 1 ] = img[:,:,1]
bimg[:,:, 2 ] = img[:,:,2]
plt.imshow(torch.cat((rimg,gimg,bimg), dim=1))
```
